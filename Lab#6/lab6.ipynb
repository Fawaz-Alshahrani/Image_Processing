```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.fft import fft2, ifft2, fftshift, ifftshift
from skimage import data, io
from skimage.transform import resize

image = data.camera()
image = resize(image, (256, 256))
io.imsave("cameraman.png", (image * 255).astype(np.uint8))

sobel_x = np.array([[-1, 0, 1],
                    [-2, 0, 2],
                    [-1, 0, 1]])
sobel_y = np.array([[-1, -2, -1],
                    [ 0,  0,  0],
                    [ 1,  2,  1]])

def center_embed_kernel(kernel, shape):
    padded = np.zeros(shape)
    kh, kw = kernel.shape
    ph, pw = shape
    cy, cx = ph // 2, pw // 2
    padded[cy - kh//2:cy - kh//2 + kh, cx - kw//2:cx - kw//2 + kw] = kernel
    return fft2(ifftshift(padded))

F = fft2(image)

H_x = center_embed_kernel(sobel_x, image.shape)
H_y = center_embed_kernel(sobel_y, image.shape)

G_x = F * H_x
G_y = F * H_y

grad_x = np.real(ifft2(G_x))
grad_y = np.real(ifft2(G_y))
magnitude = np.sqrt(grad_x**2 + grad_y**2)

plt.figure(figsize=(15, 8))
plt.subplot(2, 4, 1); plt.imshow(image, cmap='gray'); plt.title('Original Cameraman'); plt.axis('off')
plt.subplot(2, 4, 2); plt.imshow(np.log(1 + np.abs(fftshift(H_x))), cmap='gray'); plt.title('Sobel X Filter Spectrum'); plt.axis('off')
plt.subplot(2, 4, 3); plt.imshow(np.log(1 + np.abs(fftshift(H_y))), cmap='gray'); plt.title('Sobel Y Filter Spectrum'); plt.axis('off')
plt.subplot(2, 4, 4); plt.imshow(np.log(1 + np.abs(fftshift(F))), cmap='gray'); plt.title('Image Spectrum'); plt.axis('off')
plt.subplot(2, 4, 5); plt.imshow(np.log(1 + np.abs(fftshift(G_x))), cmap='gray'); plt.title('Filtered Spectrum (X)'); plt.axis('off')
plt.subplot(2, 4, 6); plt.imshow(grad_x, cmap='gray'); plt.title('Sobel X (Vertical Edges)'); plt.axis('off')
plt.subplot(2, 4, 7); plt.imshow(grad_y, cmap='gray'); plt.title('Sobel Y (Horizontal Edges)'); plt.axis('off')
plt.subplot(2, 4, 8); plt.imshow(magnitude, cmap='gray'); plt.title('Gradient Magnitude'); plt.axis('off')
plt.tight_layout()
plt.savefig("sobel_result.png", dpi=120)
plt.show()
```
